# Data Handling with NumPy & Pandas
## Objective 
- Build a strong foundation in data loading, cleaning, and manipulation using **NumPy** and **pandas**.

# Step 1: Import libraries

```python
# Step 1: Import required libraries
import pandas as pd
import numpy as np
print("Libraries imported successfully")
```

# Step 2: Load the dataset

```python
# Step 2: Load the Student Performance dataset
df = pd.read_csv("student_data.csv")

# Display confirmation message
print("Dataset loaded successfully")
print(df.head())  # View first 5 rows of the dataset
print(df.tail())  # View last 5 rows of the dataset
```

# Step 3: Data Exploration

```python
# Check dataset shape (rows, columns)
print(df.shape)
```

```python
# Display dataset information (data types & missing values)
df.info()
```

```python
# Step 7: Display column names
print(df.columns)
```

# Step 4: Data Cleaning 

```python
# Step: Check missing values in each column
missing_values = df.isnull().sum()

# Display missing values count
print("# Missing values per column")
print(missing_values)

# Check if there are any missing values in the dataset
if missing_values.sum() == 0:
    print("# No missing values found in the dataset")
else:
    print("# Missing values detected — cleaning required")
```

```python
# Step: Check for duplicate rows in the dataset
duplicate_count = df.duplicated().sum()

# Display number of duplicate rows
print(duplicate_count)

# Check if duplicates exist
if duplicate_count == 0:
    print("# No duplicate rows found in the dataset")
else:
    print("# Duplicate rows detected — removing duplicates")

    # Remove duplicate rows
    df = df.drop_duplicates()

    # Confirm duplicates are removed
    print("# Duplicate rows after removal")
    print(df.duplicated().sum())
```

# Step 5: Array operations

```python
# Step 5: Convert final grade column (G3) to NumPy array
grades = df['G3'].to_numpy()
# Step 16: Display first 5 values of NumPy array
print("# First 5 grades (NumPy array)")
print(grades[:5])
```

# Step 6: Calculate basic statistics using NumPy

```python
# Step 17: Calculate basic statistics using NumPy
mean_grade = np.mean(grades)
median_grade = np.median(grades)
std_grade = np.std(grades)

# Step 18: Display statistics
print("# Grade statistics")
print("Mean:", mean_grade)
print("Median:", median_grade)
print("Standard Deviation:", std_grade)
```
