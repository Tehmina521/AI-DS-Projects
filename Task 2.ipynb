# Step 1: Import Required Libraries

```python
# pandas → used for loading, inspecting, and manipulating tabular data (CSV files)
import pandas as pd

# matplotlib → core plotting library (low-level control over figures and axes)
import matplotlib.pyplot as plt

# seaborn → built on top of matplotlib, used for cleaner and more informative statistical plots
import seaborn as sns
print("Libraries imported successfully")
```

# Step 2: Load the Iris dataset 

```python
# Step 2: Load the Iris dataset into a pandas DataFrame
import pandas as pd  # ensure pandas is imported from Step 1

df = pd.read_csv("iris.csv")  # Load CSV file. Replace with full path if not in current folder

# Print confirmation that data is loaded
print("✅ Dataset loaded successfully.")
print(df.head())  # View first 5 rows of the dataset
```

# Step 3: Inspect dataset structure

```python
# Step 3: Inspect dataset structure

# 1️⃣ Check number of rows and columns
print("Dataset shape (rows, columns):", df.shape)

# 2️⃣ Check column names
print("Columns in dataset:", df.columns)
```

# Step 4: Visualizations

# Step 4a: Scatter Plot 
The scatter plot shows the relationship between two features (sepal length and width) and how the three species are distributed.

```python
# Step 4a: Scatter Plot to analyze relationships between variables

sns.set(style="whitegrid")  # clean style for plots

# Scatter plot: Sepal Length vs Sepal Width, colored by species
plt.figure(figsize=(7,5))  # set figure size
sns.scatterplot(
    data=df,
    x="sepal_length",  # x-axis feature
    y="sepal_width",   # y-axis feature
    hue="species",     # color points by species
    palette="Set1"     # set color palette
)
plt.title("Scatter Plot: Sepal Length vs Sepal Width")  # plot title
plt.xlabel("Sepal Length (cm)")  # x-axis label
plt.ylabel("Sepal Width (cm)")   # y-axis label
plt.legend(title="Species")      # legend with title
plt.show()
```

## Key Observations:

- Each color represents a species (setosa, versicolor, virginica).

- Setosa usually forms a distinct cluster, separating clearly from the others.

- Versicolor and Virginica may overlap slightly, indicating that these features alone may not perfectly separate them.

# Step 4b: Histogram
The histogram shows the distribution of petal length across all samples in the dataset. It helps understand how values are spread and where most of the data lies.

```python
# Step 4b: Histogram to examine data distribution

plt.figure(figsize=(7,5))  # figure size
sns.histplot(
    data=df,
    x="petal_length",  # feature to analyze
    bins=20,           # number of bins
    kde=True,          # overlay Kernel Density Estimate
    color="skyblue"    # bar color
)
plt.title("Histogram of Petal Length")  # plot title
plt.xlabel("Petal Length (cm)")         # x-axis label
plt.ylabel("Count")                     # y-axis label
plt.show()
```

## Key Observations

- Petal lengths form distinct clusters, indicating that the data is not evenly distributed.

- The cluster at the lower end (around 1–2 cm) mainly represents Setosa, while the larger petal length ranges (around 4–6 cm) correspond to Versicolor and Virginica.

- The KDE (Kernel Density Estimation) line smooths the histogram and highlights the main peaks, making the concentration of values easier to see.

# Step 4c: Box Plot
The box plot shows the spread and distribution of petal length for each species. It helps detect outliers and compare feature values across classes.

```python
# Step 4c: Box Plot to detect outliers and compare feature spread across species

plt.figure(figsize=(7,5))  # set figure size

sns.boxplot(
    data=df,
    x="species",
    y="petal_length",
    hue="species",      # explicitly assign hue
    palette="Set2",
    dodge=False,        # ensures one box per species
    legend=False        # disables extra legend
)

plt.title("Box Plot: Petal Length by Species")  # plot title
plt.xlabel("Species")                           # x-axis label
plt.ylabel("Petal Length (cm)")                 # y-axis label
plt.show()
```

## Key Observations

- Each box represents petal length distribution for a species.

- Setosa shows the smallest range and least variability, indicating very consistent petal lengths.

- Versicolor and Virginica have larger ranges and greater variability, with some outliers present.
